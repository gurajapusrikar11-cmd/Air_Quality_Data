```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import numpy as np

file_name = 'air_quality.csv'

# Create a mock air_quality.csv file if it doesn't exist to prevent errors
if not os.path.exists(file_name):
    print(f"'{file_name}' not found. Generating a mock dataset to allow execution...")
    np.random.seed(42)
    dates = pd.date_range(start="2023-01-01", periods=30, freq="D")
    locations = ["City Center", "Suburbs", "Industrial Zone"]

    data = []
    for dt in dates:
        for loc in locations:
            # Generate mock values
            pm2_5 = np.random.uniform(10, 150) if loc != "Industrial Zone" else np.random.uniform(50, 250)
            # Assign AQI Category
            if pm2_5 < 50:
                aqi_cat = 50
            elif pm2_5 < 100:
                aqi_cat = 100
            else:
                aqi_cat = 150

            data.append({
                "Date": dt.strftime('%Y-%m-%d'),
                "Location": loc,
                "PM2_5": round(pm2_5, 2),
                "Air_Quality_Category": aqi_cat
            })

    df_mock = pd.DataFrame(data)
    df_mock.to_csv(file_name, index=False)
    print(f"Mock dataset '{file_name}' created successfully.")

# Load the dataset
df = pd.read_csv(file_name)

# --- DATA PREPARATION & AGGREGATION ---

# Filter data for just the City Center
city_center_data = df[df['Location'] == 'City Center']

# Group data by location to find average pollution levels
numeric_cols = df.select_dtypes(include='number').columns
avg_pollution = df.groupby('Location')[numeric_cols].mean()

# --- THE 5 REQUIRED CHARTS ---

# Chart 1: Line Chart (Tracking Air_Quality_Category over time at City Center)
plt.figure(figsize=(8, 4))

plt.plot(city_center_data['Date'],
         city_center_data['Air_Quality_Category'],
         marker='o', color='red')

plt.title('Chart 1: Air_Quality_Category Trend at City Center')
plt.xlabel('Date')
plt.ylabel('Air Quality Index (AQI)')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Chart 2: Bar Chart (Comparing average PM 2.5 by Location)

plt.figure(figsize=(8, 4))

# Corrected column name from 'PM2.5_Level' to 'PM2_5'
avg_pollution['PM2_5'].plot(kind='bar', color='orange')

plt.title('Chart 2: Average PM2.5 Particles by Location')
plt.xlabel('Location')
plt.ylabel('Average PM2.5')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
# Chart 3: Scatter Plot (Distribution of PM2.5 levels over Date)

plt.figure(figsize=(8, 4))
# Corrected column name from 'PM2.5_Level' to 'PM2_5'
plt.scatter(df['Date'], df['PM2_5'], color='purple', s=140)
plt.title('Chart 3: PM2.5 Level Distribution by Date')
plt.xlabel('Date')
plt.ylabel('PM2.5 Level')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#char 4
import pandas as pd
import matplotlib.pyplot as plt

# Using the already loaded 'df' from previous cells
# Corrected column name from 'PM2.5_Level' to 'PM2_5'
plt.figure(figsize=(8, 4))
plt.hist(df['PM2_5'], bins=5, color='skyblue', edgecolor='black')
plt.title('Chart 4: Distribution of PM2.5 Levels')
plt.xlabel('PM2.5 Level')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
# Chart 5: Vertical Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2_5'].max().plot(kind='bar', color='blue')
plt.title('Chart 5: Max PM2.5 Recorded per Location')
plt.xlabel('Location')
plt.ylabel('Max PM2.5')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2_5'].mean().plot(kind='barh', color='blue')
plt.title('Chart: Average PM2.5 Levels by Location')
plt.xlabel('Average PM2.5 Level')
plt.ylabel('Location')
plt.show()
```
